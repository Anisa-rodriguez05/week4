## Createing a TextBlob

```python
from textblob import TextBlob
import nltk
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
# creating first textblob
wiki = TextBlob("Python is a hhigh-level, general-purpose programming language.")
```

## Part of speach tagging

```python
lala = wiki.tags
print(lala)
```

nnp = proper noun, singular 
vbz = verb
dt = determiner 
jj = adjective

## Noun Phrase Extraction

```python
teehee = wiki.noun_phrases
print(teehee)
```

## Sentiment Analysis

0 = objective 
1 = subjective

returned as named tuple 


```python
testimonial = TextBlob("Textblob is amazingly simple to use. What great fun!")
testimonial.sentiment
testimonial.sentiment.polarity
```

## Tokenization 
can break blobs into words and sentences

```python
zen = TextBlob(
    "Beautiful is better than ugly. "
    "Explicit is better than implicit. "
    "Simple is better than complex."
)
zen.words
zen.sentences
```

important note! sentence objects have similar methods as textblobs 

```python
for sentence in zen.sentences:
    print(sentence.sentiment)
```

## Words inflection and Lemmatization 
each word in textblob or sentence words is a word object (this is a subclass)

```python
sentence = TextBlob("Use 4 spaces per indentation level.")
sentence.words
#WordList(['Use', '4', 'spaces', 'per', 'indentation', 'level'])
sentence.words[2].singularize() #space
sentence.words[-1].pluralize()
```

words can be lemmatized by using lemmatize method (wtf is this)

```python
from textblob import Word
w = Word("octopi")
print(w.lemmatize())
w = Word("went")
w.lemmatize("v")  # Pass in WordNet part of speech (verb)

```
